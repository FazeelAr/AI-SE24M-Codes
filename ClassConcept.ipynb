```python
class Person:

    def __init__(self, fname, lname):
        self.fname = fname
        self.__lname = lname
    def printString(self):
        print("Name", self.__fname + " " + self.__lname)

    @property
    def fname(self):
        return self.fname

    @fname.setter
    def fname(self, n):
        if n != None and n != "":
            self.__fname = n
        else:
            self.__fname = "default"

    @property
    def lname(self):
        return self.__lname

    @lname.setter
    def lname(self, n):
        print("in setter")
        if n != None and n != "":
            self.__lname = n
        else:
            self.__lname = "default"

```

```python
class Student (Person):
    def __init__(self,fn,ln,r,c):
        #self.__name=n
        #self._fname=fn
        #self._lname=ln
        super().__init__(fn,ln)
        self.rollno=r
        self.__cgpa=c
    @property
    def rollno(self):
        return self.rollno
    @rollno.setter
    def rollno(self,rn):
        if rn!=None and rn!="":
            self.rollno=rn
        else:
            self.rollno="default"
    @property
    def cgpa(self):
        return self.__cgpa
    @cgpa.setter
    def cgpa(self,cgpa):
        if cgpa>=0.0:
            self.__cgpa=cgpa
        else:
            self.__cgpa=0.0

    def printString(self):

        #print("Name:",self._fname," ",self._lname,"RollNo:",
         #     self.__rollno,"CGPA:",self.__cgpa)
        super().printString()
        print("RollNo:",self.rollno,"CGPA:",self.__cgpa)
```
